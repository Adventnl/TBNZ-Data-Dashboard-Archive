```python
import pandas as pd
import requests
import time
import json

# Load the Excel file from the same directory as this notebook
df = pd.read_excel("tbnz_activity_with_address_1000.xlsx")

def geocode(address):
    url = "https://nominatim.openstreetmap.org/search"
    params = {
        "q": address,
        "format": "json",
        "limit": 1
    }
    
    headers = {
        "User-Agent": "tbnz-dashboard"
    }

    try:
        response = requests.get(url, params=params, headers=headers)
        data = response.json()

        if len(data) > 0:
            return float(data[0]["lat"]), float(data[0]["lon"])
        else:
            return None, None
    except:
        return None, None


result = []

for i, row in df.iterrows():
    address = row["address"]

    lat, lon = geocode(address)

    print(i, address, lat, lon)

    # 🔥 把所有字段 + 经纬度 合并
    item = row.to_dict()
    item["latitude"] = lat
    item["longitude"] = lon

    result.append(item)

    time.sleep(1)  # 必须

# 存成一个 JSON
with open("activity_with_geo.json", "w", encoding="utf-8") as f:
    json.dump(result, f, ensure_ascii=False, indent=2)

print("✅ 已生成一个完整 JSON（带经纬度）")
```

```python
import json
import random
from datetime import datetime
```

```python
cities = [
    {"city": "Auckland", "region": "Auckland Region", "lat": -36.85, "lng": 174.76},
    {"city": "Wellington", "region": "Wellington Region", "lat": -41.29, "lng": 174.78},
    {"city": "Christchurch", "region": "Canterbury", "lat": -43.53, "lng": 172.63},
]

genders = ["male", "female"]

age_ranges = [
    {"range": "18-30", "min": 18, "max": 30},
    {"range": "30-45", "min": 30, "max": 45},
    {"range": "45-60", "min": 45, "max": 60},
]

user_types = [
    "注册用户",
    "已成功入金用户",
    "完成第一笔交易用户",
    "高净值用户",
]

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
```

```python
data = []

for i in range(2000):  # 👈 生成 2000 条
    location = random.choice(cities)
    gender = random.choice(genders)
    age_group = random.choice(age_ranges)

    age = random.randint(age_group["min"], age_group["max"])
    month = random.choice(months)
    day = random.randint(1, 28)

    item = {
        "user_id": 20000 + i,
        "city": location["city"],
        "region": location["region"],
        "address": f"{random.randint(1,300)} Queen St, {location['city']}",

        "gender": gender,
        "age": age,
        "age_range": age_group["range"],

        # 🔥 全部集中在 2024
        "activity_date": f"2024-{str(random.randint(1,12)).zfill(2)}-{str(day).zfill(2)}",

        # ❌ 没有 login_flag（已经删掉）
        "login_count": random.randint(1, 10),
        "session_count": random.randint(1, 12),

        "month": month,

        # 加一点随机偏移，不然点会重叠
        "latitude": location["lat"] + (random.random() - 0.5) * 0.1,
        "longitude": location["lng"] + (random.random() - 0.5) * 0.1,

        # ✅ 新字段
        "user_type": random.choice(user_types),
    }

    data.append(item)

len(data)
```

```python
with open("activity_with_geo_2024.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print("✅ 数据已生成")
```
